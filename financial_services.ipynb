### Prepare Data for Classification

For binary classification, we need to separate our dataset into features (X) and the target variable (y). In this case, `e_signed` is our target variable, indicating whether a user completed and e-signed their loan agreement.

```python
# Define features (X) and target (y)
X = df.drop('e_signed', axis=1)
y = df['e_signed']

print(f"Shape of X: {X.shape}")
print(f"Shape of y: {y.shape}")
```

### Exploratory Data Analysis (EDA) & Business Insights

Let's delve into the data to uncover valuable financial behavioral trends and relationships as outlined in the problem description.

#### 1. Inquiry Sensitivity: Does a higher number of hard pulls (inquiries_last_month) strongly correlate with a lower probability of completing the onboarding process?

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the average e_signed for each number of inquiries_last_month
inquiry_conversion = df.groupby('inquiries_last_month')['e_signed'].mean().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(x='inquiries_last_month', y='e_signed', data=inquiry_conversion, palette='viridis')
plt.title('Conversion Rate (e_signed) by Inquiries Last Month')
plt.xlabel('Number of Inquiries Last Month')
plt.ylabel('Average e_signed (Conversion Rate)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.show()

print("Correlation between inquiries_last_month and e_signed:", df['inquiries_last_month'].corr(df['e_signed']))
```

### Predictive Modeling: Building a Classification Model

Now that we've explored the data, let's build a classification model to predict user conversion (`e_signed`). The typical steps involve:
1.  **Splitting Data:** Divide the dataset into training and testing sets.
2.  **Preprocessing:** Handle categorical features and scale numerical features (if necessary).
3.  **Model Training:** Train a machine learning model on the training data.
4.  **Model Evaluation:** Assess the model's performance on the testing data.

#### 1. Splitting Data into Training and Testing Sets

```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")

# Check the distribution of the target variable in train and test sets
print("\nTarget distribution in training set:")
print(y_train.value_counts(normalize=True))
print("\nTarget distribution in test set:")
print(y_test.value_counts(normalize=True))
```

### 3. Feature Engineering

We will create custom indicators to enrich the dataset and potentially improve model performance.

#### Total Employment Months

Combining `years_employed` and `months_employed` into a single continuous metric.

```python
# Calculate total employment months for X_train and X_test
X_train['total_employment_months'] = (X_train['years_employed'] * 12) + X_train['months_employed']
X_test['total_employment_months'] = (X_test['years_employed'] * 12) + X_test['months_employed']

# Drop original columns to avoid multicollinearity and redundancy
X_train = X_train.drop(['years_employed', 'months_employed'], axis=1)
X_test = X_test.drop(['years_employed', 'months_employed'], axis=1)

print("X_train head with new feature:\n", X_train[['total_employment_months']].head())
print("\nX_test head with new feature:\n", X_test[['total_employment_months']].head())
```

#### Account Age

Combine `personal_account_y` and `personal_account_m` to calculate how many months an applicant has held their primary personal bank account.

```python
# Calculate account age in months for X_train and X_test
X_train['account_age_months'] = (X_train['personal_account_y'] * 12) + X_train['personal_account_m']
X_test['account_age_months'] = (X_test['personal_account_y'] * 12) + X_test['personal_account_m']

# Drop original columns
X_train = X_train.drop(['personal_account_y', 'personal_account_m'], axis=1)
X_test = X_test.drop(['personal_account_y', 'personal_account_m'], axis=1)

print("X_train head with new feature:\n", X_train[['account_age_months']].head())
print("\nX_test head with new feature:\n", X_test[['account_age_months']].head())
```

#### 2. Preprocessing: Handling Categorical Features (Re-evaluation)

Now that we have engineered new features and dropped some old ones, we need to re-identify our categorical and numerical features for the `ColumnTransformer` and re-apply the preprocessing steps.

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import pandas as pd

# Drop 'entry_id' from X_train and X_test as it's an identifier and not a feature
# This needs to be done *before* identifying categorical and numerical features
X_train_cleaned = X_train.drop('entry_id', axis=1, errors='ignore')
X_test_cleaned = X_test.drop('entry_id', axis=1, errors='ignore')

# Re-identify categorical and numerical features based on the cleaned X_train
categorical_features = X_train_cleaned.select_dtypes(include=['object', 'category']).columns
numerical_features = X_train_cleaned.select_dtypes(include=['int64', 'float64']).columns

print(f"Updated Categorical features: {list(categorical_features)}")
print(f"Updated Numerical features: {list(numerical_features)}")

# Create a column transformer for one-hot encoding categorical features and scaling numerical features
# Scaling numerical features is generally a good practice for many ML models
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features),
        ('num', StandardScaler(), numerical_features) # Add StandardScaler for numerical features
    ],
    remainder='passthrough' # Should not have any remaining columns after dropping entry_id
)

# Apply the preprocessing to the training and testing data
X_train_processed = preprocessor.fit_transform(X_train_cleaned)
X_test_processed = preprocessor.transform(X_test_cleaned)

# Get feature names after preprocessing
# This involves getting names from OneHotEncoder and then appending numerical feature names
onehot_feature_names = preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_features)

# The numerical features were passed through StandardScaler, so their names remain the same
processed_feature_names = list(onehot_feature_names) + list(numerical_features)

# Convert processed arrays back to DataFrame for easier inspection and consistency
# This also ensures column names are retained for model interpretability
X_train_processed_df = pd.DataFrame(X_train_processed, columns=processed_feature_names, index=X_train_cleaned.index)
X_test_processed_df = pd.DataFrame(X_test_processed, columns=processed_feature_names, index=X_test_cleaned.index)

print(f"\nShape of X_train_processed_df after preprocessing: {X_train_processed_df.shape}")
print(f"Shape of X_test_processed_df after preprocessing: {X_test_processed_df.shape}")

print("\nFirst 5 rows of X_train_processed_df:")
display(X_train_processed_df.head())
```

#### 3. Model Training: Logistic Regression

We will now train a Logistic Regression model on our preprocessed training data. Logistic Regression is a powerful and interpretable algorithm for binary classification.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Initialize the Logistic Regression model
# Using a solver that handles L2 regularization well, like 'liblinear'
model = LogisticRegression(solver='liblinear', random_state=42)

# Train the model on the processed training data
model.fit(X_train_processed_df, y_train)

print("Logistic Regression model trained successfully!")

# Make predictions on the processed test data
y_pred = model.predict(X_test_processed_df)

# Evaluate the model
print("\nModel Evaluation:")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

#### 4. Model Analysis: ROC Curve and Feature Importances

##### a. Receiver Operating Characteristic (ROC) Curve

```python
from sklearn.metrics import roc_curve, roc_auc_score

y_pred_proba = model.predict_proba(X_test_processed_df)[:, 1]
fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba)
roc_auc = roc_auc_score(y_test, y_pred_proba)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()

print(f"AUC Score: {roc_auc:.2f}")
```

##### b. Feature Importances

```python
# For Logistic Regression, coefficients can be interpreted as feature importances
coefficients = model.coef_[0]
feature_importance_df = pd.DataFrame({
    'Feature': X_train_processed_df.columns,
    'Importance': coefficients
})

# Sort by absolute importance to see the most influential features
feature_importance_df['Absolute Importance'] = feature_importance_df['Importance'].abs()
feature_importance_df = feature_importance_df.sort_values(by='Absolute Importance', ascending=False)

plt.figure(figsize=(12, 8))
sns.barplot(x='Importance', y='Feature', data=feature_importance_df.head(15), palette='viridis')
plt.title('Top 15 Feature Importances (Logistic Regression Coefficients)')
plt.xlabel('Coefficient Value (Importance)')
plt.ylabel('Feature')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()

print("Top 15 Most Important Features:")
display(feature_importance_df.head(15).drop('Absolute Importance', axis=1))
```
